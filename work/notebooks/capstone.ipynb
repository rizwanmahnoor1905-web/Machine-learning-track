## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Assumptions

The analysis treats the information available before the prediction point as the information that would be available when prioritizing content for review. The objective is to produce a ranking for decision support rather than to claim that any particular intervention will cause an improvement in performance. The analysis also assumes that observations with insufficient or unavailable measurement data should not be interpreted as evidence of poor performance.

Features

The model uses performance and content characteristics that are available before the label window. Candidate features include relevant search, analytics, engagement, and content-level measurements from the permitted historical window. Identifier fields such as client_id and content_id are retained for grouping, joining, and identifying recommendations, but are not used as predictive features.

Missing values are handled with consideration for their meaning. Where missingness can reflect content type or measurement availability, missingness indicators are used rather than blindly replacing missing values with zero.

Label Definition

The target represents whether a content item meets the predefined decline outcome during the subsequent evaluation period. The label definition is established before model training and is kept separate from the feature window.

trend_pct and trend_direction are excluded from the feature set because the decline label is derived from these fields. Including either would allow information used to construct the target to enter the model and would constitute label leakage.

Baseline

A simple baseline ranking is evaluated alongside the ML model. The baseline provides a reference point for determining whether the model adds useful ranking performance beyond a straightforward rule or existing signal.

The model and baseline are evaluated on the same held-out data and using the same evaluation metrics.

Validation Design

The validation design reflects the intended prediction scenario rather than relying only on a random row-level split. Where future performance is being predicted, historical observations are used for training and a later period is reserved for evaluation. Client grouping is also considered so that repeated observations from the same client do not create an unrealistically easy evaluation.

The final evaluation therefore uses a defined train/test split that prevents information from the future evaluation period from being used during model development.

Leakage Checks

Several leakage checks are performed before trusting the results:

Label leakage: trend_pct and trend_direction are excluded because they contribute to the label.
Future leakage: features are checked to ensure that they do not contain information from after the prediction point.
Identifier leakage: client and content identifiers are not used as predictive features.
Window overlap: feature and label windows are checked for temporal overlap.
Availability leakage: GA4 and GSC availability are interpreted using their explicit availability flags rather than treating zero-filled unavailable measurements as genuine zero performance.
Query-window leakage: 90-day query context is used only when its time window does not overlap the label period.

These checks are intended to make the reported model performance a more honest estimate of how the ranking approach performs on unseen data.

## 5. Limitations

*What this work cannot claim.*

This analysis identifies patterns and produces a ranking of content items for review; it does not establish that any specific change to a content item will cause improved search or analytics performance.

The results are based on the available FlyRank dataset and its observation windows. Differences in client history, missing measurements, and data availability may affect the comparisons and limit how broadly the findings can be generalized.

The model's performance is evaluated on a held-out split from this dataset. Therefore, the results should not be interpreted as evidence that the model will perform identically on other websites, clients, time periods, or search environments.

The analysis also cannot claim to have predicted or reproduced Google's ranking algorithm. It models observed outcomes in the available FlyRank data.

Finally, the ranked recommendations should be treated as decision-support priorities, not guaranteed actions or outcomes. The highest-ranked items are candidates that appear worth reviewing first based on the evidence available to the model.
