## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

Clients with no search or analytics access are ranked first because the current data indicates both access sources are unavailable, making them the highest-priority cases for review

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Create a priority score, reason code, and recommended action
df["priority_score"] = df["access_profile"].map({
    "no_search_or_analytics_access": 100,
    "gsc_only": 50,
    "ga4_only": 50,
    "FULL_ACCESS": 0
})

df["reason_code"] = df["access_profile"].map({
    "no_search_or_analytics_access": "NO_SEARCH_ANALYTICS_ACCESS",
    "gsc_only": "GA4_MISSING",
    "ga4_only": "GSC_MISSING",
    "FULL_ACCESS": "FULL_ACCESS"
})

df["action"] = df["access_profile"].map({
    "no_search_or_analytics_access": "Connect GSC and GA4",
    "gsc_only": "Connect GA4",
    "ga4_only": "Connect GSC",
    "FULL_ACCESS": "No access action needed"
})

# Rank highest-priority clients first
ranked_queue = (
    df[
        [
            "client_hash_id",
            "access_profile",
            "priority_score",
            "reason_code",
            "action"
        ]
    ]
    .sort_values(
        by=["priority_score", "client_hash_id"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

# Add queue rank
ranked_queue.insert(0, "rank", ranked_queue.index + 1)

ranked_queue.head(20)
```

Introduction / Problem

Effective content and SEO decisions depend on access to the data needed to understand search visibility and website performance. In a FlyRank client workflow, differences in access to Google Search Console (GSC) and Google Analytics 4 (GA4) can therefore create different starting points for content analysis and recommendation.

The practical problem is that clients do not necessarily have the same level of access. A client may have access to neither platform, only GSC, only GA4, or both. Treating these cases identically makes it harder to determine which access-related action should be prioritized.

This work frames the problem as a client-level prioritization task. Using the available access-profile information, clients are categorized according to their GSC and GA4 availability and assigned an interpretable score and reason code. The score is used to prioritize missing-access cases, while the reason code and action label explain what should be addressed.

The goal is not to claim that the scoring system causes better content performance. Instead, it provides a transparent decision-support mechanism that converts an existing client attribute into an actionable workflow signal. The findings therefore connect the analysis directly to the FlyRank content problem: identifying where missing search and analytics access may need to be addressed before or alongside data-driven content and SEO work.
