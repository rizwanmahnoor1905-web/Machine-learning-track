## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Fixed rules like "if amount > $5000, flag as fraud" fail because:

Legitimate spending is messy. Some people spend $5000 on vacations. Some shop at 3am. Rules can't tell the difference.
Fraudsters are smart. They know the rules. They make small transactions and mimic normal behavior. Rules don't adapt.
Context matters. $5000 is suspicious for someone who spends $100/week. Normal for someone who spends $10,000/week. Rules can't be context-dependent.
Combinations matter. $5000 alone is okay. 3am alone is okay. But $5000 at 3am from a new device in a different country? That's different. Rules don't combine signals.
Patterns are too complex. You'd need 100+ if-statements. And they'd still fail.

Why ML works:
ML learns from millions of real transactions. It finds the actual patterns — the messy, context-dependent, subtle combinations that predict fraud. It adapts per customer. It catches what rules miss
